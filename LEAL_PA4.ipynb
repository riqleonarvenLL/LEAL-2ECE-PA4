A. Visayan Communication Dataframe

```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_excel("board2.xlsx")
```

```python
# Compute Average column
df["Average"] = df[["Math","GEAS","Electronics","Communication"]].mean(axis=1)
```

```python
VisComm = df[(df["Hometown"]=="Visayas") & (df["Track"]=="Communication")]
VisComm = VisComm[["Name","Gender","Math","Electronics","Average"]]
```

```python
print("Number of rows:", len(VisComm))
VisComm
```

B. Visayan Female DataFrame

```python
# B. VisFemale DataFrame
VisFemale = df[(df["Hometown"]=="Visayas") & (df["Gender"]=="Female")]
VisFemale = VisFemale[["Name","Track","GEAS","Electronics","Average"]]
```

C. CATEGORY-AVERAGE VISUALIZATION

```python
import matplotlib.pyplot as plt
```

```python
# Category-Average Visualization
avg_by_track = df.groupby("Track")["Average"].mean()
avg_by_gender = df.groupby("Gender")["Average"].mean()
avg_by_home = df.groupby("Hometown")["Average"].mean()
```

```python
# Display summary tables
print("Mean Average by Track:\n", avg_by_track, "\n")
print("Mean Average by Gender:\n", avg_by_gender, "\n")
print("Mean Average by Hometown:\n", avg_by_home, "\n")
```

```python

# Plot three bar charts in one figure
fig, axes = plt.subplots(1,3,figsize=(15,5))
```

```python
avg_by_track.plot(kind="bar", ax=axes[0], title="Mean Average by Track")
avg_by_gender.plot(kind="bar", ax=axes[1], title="Mean Average by Gender")
avg_by_home.plot(kind="bar", ax=axes[2], title="Mean Average by Hometown")
```

```python
for ax in axes:
    ax.set_ylabel("Average Score")
    ax.set_xlabel("Category")
```

```python
plt.tight_layout()
plt.show()
```

```python
# Interpretation
print("Track with highest mean:", avg_by_track.idxmax())
print("Gender with highest mean:", avg_by_gender.idxmax())
print("Hometown with highest mean:", avg_by_home.idxmax())
```
